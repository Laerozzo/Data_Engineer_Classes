## Chapter 1: Foundations for Efficiencies

### 1.1 Defining "Efficient" and "Pythonic" Code
* **Efficient Code**: Code that executes with **minimal completion time** (fast runtime) and **minimal resource consumption** (small memory footprint).
* **Pythonic Code**: Code that adheres to idioms intended by Python's creators, prioritizing readability, conciseness, and native constructs (e.g., list comprehensions `[x * 2 for x in numbers]` instead of manual index-based `for` loops).
* **The Zen of Python (`import this`)**: Guiding aphorisms by Tim Peters emphasizing that *"Beautiful is better than ugly"*, *"Simple is better than complex"*, *"Flat is better than nested"*, and *"Readability counts"*.

### 1.2 Building with Python Built-ins
* Every standard Python installation includes the **Python Standard Library**:
  * **Built-in Types**: `list`, `tuple`, `set`, `dict`, etc.
  * **Built-in Functions**: `print()`, `len()`, `range()`, `round()`, `enumerate()`, `map()`, `zip()`, etc.
  * **Built-in Modules**: `os`, `sys`, `itertools`, `collections`, `math`, etc.
* **Key Built-in Iterators** (return lazy iterator objects that can be unpacked with `[*iterator]` or `list(iterator)`):
  * `range(start, stop, step)`: Generates sequences of integers without materializing a list in memory until needed.
  * `enumerate(iterable, start=0)`: Creates an indexed sequence of `(index, value)` tuples.
  * `map(function, iterable)`: Applies a function (built-in or `lambda`) across every element of an iterable.

### 1.3 The Power of NumPy Arrays
* **Homogeneity**: Unlike Python lists, `np.array()` enforces a single data type (`dtype` such as `int64` or `float64`) across all elements, enabling contiguous memory storage and C-level speed.
* **Broadcasting (Vectorization)**: Python lists raise a `TypeError` on element-wise arithmetic (`nums ** 2`), requiring slow loops or list comprehensions. NumPy arrays broadcast operations across all elements simultaneously (`nums_np ** 2`).
* **2-D Indexing**: Slice rows and columns cleanly with `nums2_np[row_idx, col_idx]` (e.g., `nums2_np[:, 0]`) instead of nested list comprehensions.
* **Boolean Indexing**: Filter arrays in one step using boolean masks (`nums_np[nums_np > 0]`).

```python
# The Zen of Python & Non-Pythonic vs. Pythonic Comparison
import this

numbers = [1, 2, 3, 4, 5]

# Non-Pythonic approach
doubled_numbers_loop = []
for i in range(len(numbers)):
    doubled_numbers_loop.append(numbers[i] * 2)

# Pythonic list comprehension approach
doubled_numbers_pythonic = [x * 2 for x in numbers]
print("\nDoubled numbers:", doubled_numbers_pythonic)
```

```python
# Building with Built-ins: range(), enumerate(), and map()
# 1. range(start, stop) and range(start, stop, step)
nums = range(0, 11)
print("range(0, 11):", list(nums))

even_nums = range(2, 11, 2)
print("range(2, 11, 2):", [*even_nums])

# 2. enumerate() with default and custom start index
letters = ['a', 'b', 'c', 'd']
indexed_letters = enumerate(letters)
print("enumerate default:", list(indexed_letters))

indexed_letters2 = enumerate(letters, start=5)
print("enumerate(start=5):", list(indexed_letters2))

# 3. map() with built-in round and with lambda
float_nums = [1.5, 2.3, 3.4, 4.6, 5.0]
rnd_nums = map(round, float_nums)
print("map(round):", list(rnd_nums))

int_nums = [1, 2, 3, 4, 5]
sqrd_nums = map(lambda x: x ** 2, int_nums)
print("map(lambda):", list(sqrd_nums))
```

```python
# The Power of NumPy Arrays: Homogeneity, Broadcasting, 2-D Slicing & Boolean Indexing
import numpy as np

# 1. Array homogeneity
nums_np_ints = np.array([1, 2, 3])
nums_np_floats = np.array([1, 2.5, 3])
print("Ints dtype:", nums_np_ints.dtype, "| Upcast floats dtype:", nums_np_floats.dtype)

# 2. Broadcasting vs list comprehension
nums = [-2, -1, 0, 1, 2]
nums_np = np.array(nums)
print("Vectorized squaring:", nums_np ** 2)

# 3. 2-D Indexing
nums2 = [[1, 2, 3], [4, 5, 6]]
nums2_np = np.array(nums2)
print("Element (0, 1):", nums2_np[0, 1])
print("First column:", nums2_np[:, 0])

# 4. Boolean Indexing
boolean_mask = nums_np > 0
print("Boolean mask:", boolean_mask)
print("Filtered positive values:", nums_np[boolean_mask])
```

## Chapter 2: Timing and Profiling Code

### 2.1 Examining Runtime with `%timeit`
* **IPython Magic Commands**: Enhancements prefixed with `%` (line magic for a single line) or `%%` (cell magic for multiple lines). Run `%lsmagic` to list all available magic commands.
* **Using `%timeit` and `%%timeit`**: Runs a statement multiple times across multiple runs and reports the mean and standard deviation of execution time:
  * **Customizing Runs & Loops**: Use `-r` to set number of runs and `-n` to set number of loops per run (e.g., `%timeit -r2 -n10 rand_nums = np.random.rand(1000)`).
  * **Saving Output (`-o`)**: `times = %timeit -o ...` stores a `TimeitResult` object exposing `.timings` (list of run times), `.average`, `.best`, and `.worst`.
* **Literal vs. Formal Syntax**: Creating data structures via literal syntax (`[]`, `{}`, `()`) is significantly faster than calling formal constructors (`list()`, `dict()`, `tuple()`).

### 2.2 Code Profiling for Runtime (`line_profiler`)
* Timing an entire function with `%timeit` doesn't show *which internal line* is the bottleneck.
* **`line_profiler` Package (`pip install line_profiler`)**:
  1. Load extension: `%load_ext line_profiler`
  2. Profile function line-by-line: `%lprun -f func_name func_name(arg1, arg2)`
* **Output Columns**: `Line #`, `Hits` (number of times that line executed), `Time` (total time spent on that line in timer units), `Per Hit` (average time per execution), `% Time` (percentage of total function runtime spent on that line), and `Line Contents`.

### 2.3 Code Profiling for Memory Usage (`sys.getsizeof` & `memory_profiler`)
* **Quick & Dirty Check**: `import sys; sys.getsizeof(obj)` returns the size of an individual object in bytes.
* **Line-by-Line Memory Profiling (`pip install memory_profiler`)**:
  1. **Important Caveat**: Functions profiled with `%mprun` **must be defined in an external `.py` module** and imported into the session (e.g., `from hero_funcs import convert_units`), not defined interactively in the notebook cell.
  2. Load extension: `%load_ext memory_profiler`
  3. Run profiler: `%mprun -f convert_units convert_units(heroes, hts, wts)`
* **Output Caveats**: Reports `Mem usage` and `Increment` in `MiB` (Mebibytes). Small allocations (e.g., 480 rows) may register as `0.0 MiB`, so profiling on larger samples (e.g., 35,000 rows) provides clearer line-by-line comparisons.

```python
# Timing code with Python's timeit module (and demonstrating IPython %timeit equivalents)
import timeit
import numpy as np

# Comparing formal constructor dict() vs literal syntax {}
f_time = timeit.timeit("formal_dict = dict()", number=1000000)
l_time = timeit.timeit("literal_dict = {}", number=1000000)
diff_ns = ((f_time - l_time) / 1000000) * (10 ** 9)

print(f"Formal dict() avg: {(f_time / 1e6) * 1e9:.1f} ns")
print(f"Literal {{}} avg:   {(l_time / 1e6) * 1e9:.1f} ns")
print(f"Literal syntax is faster by ~{diff_ns:.1f} ns per call")

# In an interactive IPython / Jupyter cell, you can also run:
# %timeit -r2 -n10 rand_nums = np.random.rand(1000)
# times = %timeit -o rand_nums = np.random.rand(1000)
```

## Chapter 3: Gaining Efficiencies

### 3.1 Efficiently Combining, Counting, and Iterating
* **Combining with `zip()`**: Instead of looping with `enumerate()` and manual indexing, `zip(iter1, iter2)` pairs elements into tuples lazily (unpack via `[*zip(names, hps)]`).
* **`collections` Module** (Specialized Container Datatypes):
  * `Counter(iterable)`: Dictionary subclass that counts hashable objects in a single C-optimized pass without manual `if key not in dict:` loops.
  * Other notable containers: `namedtuple` (tuple subclasses with named fields), `deque` (fast appends/pops on both ends), `OrderedDict`, `defaultdict`.
* **`itertools` Module** (Functional Iterator Tools):
  * **Combination Generators**: `combinations(iterable, r)` (unordered unique pairs), `permutations`, `product`.
  * **Infinite Iterators**: `count`, `cycle`, `repeat`.
  * **Finite Iterators**: `accumulate`, `chain`, `zip_longest`.

### 3.2 Set Theory for Fast Comparisons & Membership Testing
* Python's built-in `set` datatype supports mathematical set operations that eliminate nested comparison loops:
  * `set_a.intersection(set_b)`: Elements present in **both** `set_a` and `set_b`.
  * `set_a.difference(set_b)`: Elements in `set_a` **but not** in `set_b`.
  * `set_a.symmetric_difference(set_b)`: Elements in **exactly one** of the sets (not both).
  * `set_a.union(set_b)`: All unique elements across **either** set.
* **Fast Membership Testing (`in`)**: Checking `'Zubat' in names_set` uses hash-table lookup ($O(1)$ time, ~37 ns) and is hundreds of times faster than scanning a `list` or `tuple` ($O(n)$ time, ~7.6 $\mu$s).
* **Unique Extraction**: `set(primary_types)` extracts unique items instantly compared to a `for` loop with `if item not in unique_list:`.

### 3.3 Eliminating Loops & Writing Better Loops
* **Eliminating Loops**:
  * Replace row-summing `for` loops over 2-D lists with `[sum(row) for row in poke_stats]` or `[*map(sum, poke_stats)]`.
  * Replace row-averaging loops over 2-D arrays with vectorized axis methods: `poke_stats.mean(axis=1)`.
* **Writing Better Loops (When Looping is Unavoidable)**:
  * **Move One-Time Calculations Above the Loop**: If `total_attack_avg = attacks.mean()` doesn't change per iteration, compute it **once** before the loop begins.
  * **Use Holistic Conversions Below the Loop**: Instead of calling `list(poke_tuple)` inside every iteration, collect tuples in the loop and convert all at once outside the loop via `[*map(list, poke_data_tuples)]`.

```python
# Combining with zip(), counting with collections.Counter(), and generating pairs with itertools.combinations()
import pandas as pd
from collections import Counter
from itertools import combinations

pokemon_df = pd.read_csv('datasets/pokemon.csv')

# 1. Combining objects with zip()
names = pokemon_df['Name'][:3].tolist()
hps = pokemon_df['HP'][:3].tolist()
combined_zip = zip(names, hps)
combined_zip_list = [*combined_zip]
print("Combined with zip():", combined_zip_list)

# 2. Counting types with collections.Counter()
poke_types = pokemon_df['Primary_Type'].tolist()
type_counts = Counter(poke_types)
print("\nType counts with Counter:", type_counts)

# 3. Generating type combinations with itertools.combinations()
sample_types = ['Bug', 'Fire', 'Ghost', 'Grass', 'Water']
combos_obj = combinations(sample_types, 2)
combos = [*combos_obj]
print("\n2-Type combinations (10 total):", combos)
```

```python
# Set Theory Operations & Membership Testing
list_a = ['Bulbasaur', 'Charmander', 'Squirtle']
list_b = ['Caterpie', 'Pidgey', 'Squirtle']

set_a = set(list_a)
set_b = set(list_b)

print("Intersection (in both):         ", set_a.intersection(set_b))
print("Difference (in A, not B):       ", set_a.difference(set_b))
print("Difference (in B, not A):       ", set_b.difference(set_a))
print("Symmetric Difference (in 1 set):", set_a.symmetric_difference(set_b))
print("Union (all unique in A or B):   ", set_a.union(set_b))

# Membership check & unique extraction
names_set = set(pokemon_df['Name'])
print("\nIs 'Zubat' in names_set?", 'Zubat' in names_set)
print("Unique Primary Types:", set(pokemon_df['Primary_Type']))
```

```python
# Eliminating Loops with Built-ins / NumPy & Writing Better Loops
import numpy as np

poke_stats_list = pokemon_df[['HP', 'Attack', 'Defense', 'Speed']].values.tolist()
poke_stats_np = pokemon_df[['HP', 'Attack', 'Defense', 'Speed']].values

# 1. Eliminating row-sum loops with map()
totals_map = [*map(sum, poke_stats_list)]
print("Row totals via [*map(sum, ...)]:", totals_map[:5])

# 2. Eliminating row-mean loops with NumPy .mean(axis=1)
avgs_np = poke_stats_np.mean(axis=1)
print("Row averages via .mean(axis=1): ", avgs_np[:5])

# 3. Writing Better Loops: Moving one-time calculation ABOVE the loop
sub_names = ['Absol', 'Aron', 'Jynx', 'Natu', 'Onix']
attacks = np.array([130, 70, 50, 50, 45])

# Calculate mean ONCE outside the loop
total_attack_avg = attacks.mean()
print("\nPokémon above average attack:")
for pokemon, attack in zip(sub_names, attacks):
    if attack > total_attack_avg:
        print(f"  {pokemon}'s attack: {attack} > average: {total_attack_avg}!")

# 4. Writing Better Loops: Holistic conversion BELOW the loop
poke_data_tuples = []
for poke_tuple in zip(pokemon_df['Name'][:3], pokemon_df['Legendary'][:3], pokemon_df['Generation'][:3]):
    poke_data_tuples.append(poke_tuple)

# Convert tuples to lists holistically after the loop
poke_data = [*map(list, poke_data_tuples)]
print("\nHolistically converted list of lists:", poke_data)
```

#### Key Takeaways
* Standard library tools (`zip()`, `collections.Counter()`, `itertools.combinations()`) replace verbose, error-prone loops with fast C-backed iterators.
* Python `set` methods (`intersection`, `difference`, `symmetric_difference`, `union`) and `in` membership checks operate in near-constant $O(1)$ time per element.
* Loops over 2-D numerical data can be eliminated completely using `[*map(sum, data)]` or `np_array.mean(axis=1)`.

3. **Collapse 2-D Aggregations**: Use `np_array.mean(axis=1)` or `np_array.sum(axis=1)` instead of iterating row-by-row.
4. **Hoist Invariants Out of Loops**: Inspect every line inside remaining `for` loops—if an expression doesn't depend on the loop variable, move it above the loop.

## Chapter 4: Basic `pandas` Optimizations

### 4.1 Intro to `pandas` DataFrame Iteration (`.iloc`, `.iterrows()`, `.itertuples()`)
* `pandas` is built on top of `NumPy` arrays and stores tabular data with labeled rows and columns.
* **1. Iterating with `.iloc[i]` (Anti-Pattern — Slowest: ~183 ms)**:
  * Looping `for i in range(len(baseball_df)): row = baseball_df.iloc[i]` performs repeated positional lookups and creates a `Series` object on every pass.
* **2. Iterating with `.iterrows()` (Better: ~95.3 ms)**:
  * `for i, row in baseball_df.iterrows():` yields `(index, Series)` pairs. About 2x faster than `.iloc`, but still incurs overhead from casting each row into a `pandas.Series` (access values via `row['W']`).
* **3. Iterating with `.itertuples()` (Best Row Iterator: ~7.48 ms)**:
  * `for row_namedtuple in team_wins_df.itertuples():` yields `namedtuple` objects (`Pandas(Index=0, Team='ARI', Year=2012, W=81)`).
  * Over **70x faster** than `.iterrows()` because `namedtuple` avoids `Series` type-checking overhead.
  * **Crucial Caveat**: You **must** access attributes using **dot notation** (`row_namedtuple.Team`, `row_namedtuple.Index`); bracket indexing (`row_namedtuple['Team']`) raises `TypeError: tuple indices must be integers or slices, not str`.

### 4.2 `pandas` Alternative to Looping: `.apply()` (~30.1 ms)
* `baseball_df.apply(lambda row: calc_run_diff(row['RS'], row['RA']), axis=1)` takes a function and applies it across columns (`axis=0`) or rows (`axis=1`).
* Eliminates explicit `for` loop boilerplate and runs ~3x faster than `.iterrows()`.

### 4.3 Optimal `pandas` Iterating: NumPy Vectorization (~124 $\mu$s)
* Because `pandas` builds on `NumPy`, extracting underlying column arrays via `.values` (e.g., `baseball_df['W'].values`, which returns `<class 'numpy.ndarray'>`) enables **broadcasting**:
  `run_diffs_np = baseball_df['RS'].values - baseball_df['RA'].values`
* Runs in **124 $\mu$s**—over **240x faster** than `.apply()` and **700x faster** than `.iterrows()`!

```python
# Comparing DataFrame Iteration: .iloc vs .iterrows() vs .itertuples()
import numpy as np
import pandas as pd

baseball_df = pd.read_csv('datasets/baseball_stats.csv')

def calc_win_perc(wins, games_played):
    win_perc = wins / games_played
    return np.round(win_perc, 2)

# 1. Iterating with .iterrows()
win_perc_list = []
for i, row in baseball_df.iterrows():
    wins = row['W']
    games_played = row['G']
    win_perc = calc_win_perc(wins, games_played)
    win_perc_list.append(win_perc)

baseball_df['WP'] = win_perc_list
print("Baseball DataFrame with Win Percentage (WP):")
print(baseball_df.head())

# 2. Iterating with .itertuples() using dot notation
print("\nIterating with .itertuples() (first 3 rows):")
for row_namedtuple in baseball_df[['Team', 'Year', 'W']].head(3).itertuples():
    print(f"  Index={row_namedtuple.Index}, Team={row_namedtuple.Team}, W={row_namedtuple.W}")
```

```python
# Calculating Run Differentials (RD): .iterrows() vs .apply() vs NumPy Vectorization (.values)
import timeit

def calc_run_diff(runs_scored, runs_allowed):
    return runs_scored - runs_allowed

# 1. Using .apply() with axis=1
run_diffs_apply = baseball_df.apply(
    lambda row: calc_run_diff(row['RS'], row['RA']),
    axis=1
)
baseball_df['RD_apply'] = run_diffs_apply

# 2. Optimal Approach: NumPy array broadcasting with .values
run_diffs_np = baseball_df['RS'].values - baseball_df['RA'].values
baseball_df['RD'] = run_diffs_np

print("Baseball DataFrame with Run Differential (RD):")
print(baseball_df[['Team', 'League', 'Year', 'RS', 'RA', 'W', 'Playoffs', 'WP', 'RD']].head())

# Benchmark the 3 approaches over 1,000 runs
t_iterrows = timeit.timeit(
    stmt="[calc_run_diff(row['RS'], row['RA']) for _, row in baseball_df.iterrows()]",
    globals=globals(), number=500
)
t_apply = timeit.timeit(
    stmt="baseball_df.apply(lambda row: calc_run_diff(row['RS'], row['RA']), axis=1)",
    globals=globals(), number=500
)
t_values = timeit.timeit(
    stmt="baseball_df['RS'].values - baseball_df['RA'].values",
    globals=globals(), number=500
)

print(f"\nBenchmark (500 loops):")
print(f"  .iterrows() total: {t_iterrows*1000:.2f} ms")
print(f"  .apply() total:    {t_apply*1000:.2f} ms")
print(f"  .values array:     {t_values*1000:.2f} ms (Fastest!)")
```

### Chapter 4: Key Takeaways & Actionable Step-by-Step Process

#### Key Takeaways
* Performance hierarchy for `pandas` DataFrame operations (from slowest to fastest):
  $$	ext{.iloc loop} \ll 	ext{.iterrows()} \ll 	ext{.apply(axis=1)} \ll 	ext{.itertuples()} \ll 	ext{NumPy .values Vectorization}$$
* When sequential row iteration is strictly required, always choose `.itertuples()` over `.iterrows()` and reference column fields using dot notation (`row.Team`).
* For arithmetic or mathematical transformations across columns, bypass row iteration entirely by operating on underlying NumPy arrays (`df['ColA'].values - df['ColB'].values`).

#### Step-by-Step Process to Act On
1. **Check for Vectorization First**: Ask if the calculation can be expressed as array arithmetic on `df['col'].values` (or Series arithmetic `df['ColA'] - df['ColB']`).
2. **Use `.itertuples()` for Sequential Row Logic**: If you must loop over rows, use `for row in df.itertuples():` and access attributes via `row.ColName`.
3. **Use `.apply()` for Complex Row/Column Functions**: When applying a custom function that cannot be easily vectorized, call `df.apply(lambda row: ..., axis=1)` (for rows) or `axis=0` (for columns).
4. **Never Loop with `.iloc[i]`**: Eliminate any `for i in range(len(df)): df.iloc[i]` patterns immediately.
