## Chapter 1: Introduction to Python

### 1.1 Basic Syntax, Comments, and Variables
* **Comments**: Lines starting with `#` are ignored by the Python interpreter and used to document code.
* **Variable Naming Conventions**: Use `snake_case` (all lowercase letters with underscores `_` separating words).
* **Variable Assignment**: The `=` operator assigns a value to a variable name (`variable_name = value`).
* **Basic Data Types**:
  * `str` (String): Text enclosed in single (`'...'`) or double (`"..."`) quotation marks. Double quotes are recommended when strings contain apostrophes (e.g., `"Chef's recipe"`). Multi-line strings use triple quotes (`"""..."""`).
  * `int` (Integer): Whole numbers without decimals (no quotation marks).
  * `float` (Floating-Point): Numbers with decimal points.
  * `bool` (Boolean): Logical truth values (`True` or `False` — case-sensitive).

### 1.2 Operators and String Methods
* **Arithmetic Operators**: `+` (addition), `-` (subtraction), `*` (multiplication), `/` (division).
* **String Operators**: `+` concatenates strings together; `*` repeats a string $n$ times.
* **String Methods**: Functions bound specifically to `str` objects, called via dot notation (`string_var.method()`):
  * `.replace("old", "new")`: Replaces occurrences of a substring with a new substring.
  * `.lower()` / `.upper()`: Converts all characters to lowercase or uppercase.

## Chapter 2: Working with Data Structures

### 2.1 Lists
* **Overview**: Ordered, mutable collections enclosed in square brackets `[]` that can store multiple values of mixed data types.
* **Indexing**: Zero-based (`0` is the first element). Negative indices count from the end (`-1` is the last element, `-2` is second-to-last).
* **Slicing (`list[start:stop:step]`)**:
  * `list[1:3]`: Extracts elements from index `1` up to (but **not including**) index `3`.
  * `list[3:]`: Extracts all elements from index `3` to the end.
  * `list[:3]`: Extracts all elements from the beginning up to index `2` (excludes index `3`).
  * `list[::2]`: Extracts every second element across the entire list.

### 2.2 Dictionaries
* **Overview**: Ordered, mutable mappings of `key: value` pairs enclosed in curly brackets `{}`. Keys must be unique and immutable.
* **Accessing & Updating**: Retrieve values via `dict["key"]`. Add or overwrite a pair via `dict["key"] = value`.
* **Dictionary Methods**:
  * `.keys()`: Returns a view object of all dictionary keys.
  * `.values()`: Returns a view object of all dictionary values.
  * `.items()`: Returns a view object of `(key, value)` tuples.

### 2.3 Sets and Tuples
* **Sets (`{val1, val2}` or `set(iterable)`)**: Unordered, unindexed collections of **unique** elements. Duplicate values are automatically removed. Elements cannot be updated in place, though items can be added or removed. Use `sorted(set_obj)` to return an alphabetically/numerically sorted list.
* **Tuples (`(val1, val2)`)**: Ordered, indexed, and **immutable** collections enclosed in parentheses `()`. Once created, elements cannot be added, removed, or modified.

```python
# Create a list of ingredients and quantities
ingredients = ["fusilli", "tomatoes", "garlic", "basil", "olive oil", "salt"]
quantities = [500, 400, 15, 20, 30, 10]

print(ingredients)
print(quantities)

# Get the second ingredient (index 1)
second_ingredient = ingredients[1]
print("Second ingredient:", second_ingredient)

# Extract the last quantity using negative indexing
last_quantity = quantities[-1]
print("Last quantity:", last_quantity)

# Get every other ingredient starting from the first
alternate_ingredient = ingredients[::2]
print("Alternate ingredients:", alternate_ingredient)
```

```python
# Create the recipe dictionary
recipe = {
    "olive oil": 30,
    "garlic": 15,
    "tomatoes": 400
}
print(recipe)

# Add basil to the recipe dictionary
recipe["basil"] = 20
print(recipe)

# Retrieve keys, values, and items
ingredient_names = recipe.keys()
quantities = recipe.values()
recipe_items = recipe.items()

print("Ingredient names:", ingredient_names)
print("Quantities:", quantities)
print("Recipe items:", recipe_items)
```

```python
# Create a tuple (immutable)
cup_conversion = (1, 240)
print(type(cup_conversion))

# Convert ingredients list to a set to enforce uniqueness
unique_ingredients = set(ingredients)
print(unique_ingredients)

# Sort unique_ingredients alphabetically (returns a list)
print(sorted(unique_ingredients))
```

### Chapter 2: Key Takeaways & Actionable Step-by-Step Process

#### Key Takeaways
* Choose the right data structure based on mutability, ordering, and lookup needs:
  * **List (`[]`)**: Ordered sequence for general-purpose iteration and slicing.
  * **Dictionary (`{k: v}`)**: Fast key-based lookup for structured records.
  * **Set (`{v}`)**: Instant deduplication and membership testing.
  * **Tuple (`()`)**: Fixed, write-protected records (e.g., coordinates, conversion constants).

#### Step-by-Step Process to Act On
1. **Assess Data Needs**: Determine whether your collection requires key-value mapping (`dict`), uniqueness (`set`), immutability (`tuple`), or sequential indexing (`list`).
2. **Deduplicate Raw Lists**: Pass raw lists through `set()` and `sorted()` to get a clean, ordered list of distinct categories.
3. **Structure Related Attributes**: Combine parallel lists (like `ingredients` and `quantities`) into a single dictionary for safer, key-driven lookups.

## Chapter 3: Control Flow and Loops

### 3.1 Comparison Operators & Conditional Statements
* **Comparison Operators**: Evaluate relationships and return a `bool` (`True` or `False`):
  * `==` (equal to), `!=` (not equal to), `<` (less than), `<=` (less than or equal to), `>` (greater than), `>=` (greater than or equal to).
* **Conditional Statements (`if` / `elif` / `else`)**:
  * Every clause header ends with a colon `:`, and the body beneath it **must be indented** (standard is 4 spaces).
  * `elif` checks an alternative condition only if all preceding `if`/`elif` conditions evaluated to `False`.
  * `else` acts as a catch-all fallback when all prior conditions are `False`.

### 3.2 `for` Loops and `range()`
* **Syntax**: `for item in iterable:` iterates over each element in a sequence (list, dictionary, string, tuple).
* **Iterating Dictionaries**: Use `for key, val in dict.items():` to unpack keys and values simultaneously.
* **`range(start, stop, step)`**: Generates a sequence of numbers from `start` up to `stop - 1`.

### 3.3 `while` Loops and Workflow Keywords
* **`while` Loop**: Repeats an action as long as a condition evaluates to `True`. Always update the state variable inside the loop (e.g., `counter += 1`) or use `break` to prevent an infinite loop.
* **Membership & Logical Operators**:
  * `in` / `not in`: Checks whether an element exists inside a collection.
  * `and`: Returns `True` only if **both** conditions are `True`.
  * `or`: Returns `True` if **at least one** condition is `True`.

```python
# Conditional statement example: checking pantry stock
pantry_stock = {"tomatoes": 200, "fusilli": 300, "olive oil": 50}
ingredients_needed = {"tomatoes": 1000}

if pantry_stock["tomatoes"] >= ingredients_needed["tomatoes"]:
    print("Enough tomatoes for the party!")
elif pantry_stock["tomatoes"] >= 800:
    print("Only enough tomatoes for a smaller gathering.")
else:
    print("Need to buy tomatoes before the party.")
```

```python
# Looping through a list and categorizing quantities
ingredients = ["fusilli", "tomatoes", "garlic", "basil", "olive oil", "salt"]
for item in ingredients:
    print(item)

print("---")
quantities = [500, 400, 20, 15, 15, 7]
for qty in quantities:
    if qty >= 400:
        print("Large quantity")
    elif qty >= 200:
        print("Medium quantity")
    else:
        print("Small quantity")
```

```python
# Looping through dictionary items to scale a recipe
recipe = {
    "fusilli": 500,
    "tomatoes": 400,
    "basil": 20,
    "garlic": 15,
    "olive oil": 15,
    "salt": 7
}

for ingredient, qty in recipe.items():
    scaled_qty = qty * 2
    print(ingredient, ":", scaled_qty, "g")
```

```python
# While loop examples
total_confirmations = 10
guest_count = 0

while guest_count < total_confirmations:
    guest_count += 1
    print(guest_count, "guests so far!")
print("We have", guest_count, "guests coming!")

print("---")
total_ingredients = 7
ingredients_checked = 0

while ingredients_checked < total_ingredients:
    ingredients_checked += 1
    if ingredients_checked < 4:
        print("More than half remaining")
    elif ingredients_checked <= 6:
        print("Nearly finished checking")
    else:
        print("All ingredients verified!")
```

```python
# Building a safe workflow: Avoiding KeyError when checking dictionaries
# Note: Directly accessing pantry_stock[ingredient] when 'ingredient' is missing raises a KeyError!
# Solution: Check membership with 'not in' BEFORE indexing the dictionary.

pantry = {"tomatoes": 200, "fusilli": 600, "salt": 50}
scale_factor = 2
shopping_list = []

for ingredient, amount in recipe.items():
    needed_amount = amount * scale_factor
    # Short-circuit evaluation prevents KeyError if ingredient is not in pantry
    if ingredient not in pantry or needed_amount > pantry[ingredient]:
        shopping_list.append(ingredient)

print("Shopping list for your party:", shopping_list)
```

### Chapter 3: Key Takeaways & Actionable Step-by-Step Process

#### Key Takeaways
* `if`/`elif`/`else` blocks direct program execution based on boolean expressions.
* `for` loops are ideal when the number of iterations is known (iterating over a collection or `range()`), whereas `while` loops are best when repeating until a dynamic condition changes.
* Combining `not in` with `or` protects dictionary lookups from raising a `KeyError` on missing keys.

#### Step-by-Step Process to Act On
1. **Initialize Accumulators**: Create empty lists (`shopping_list = []`) or counters (`count = 0`) before entering a loop.
2. **Select Loop Type**: Use `for k, v in dict.items():` for key-value processing, or `while` when waiting for a threshold to be met.
3. **Guard Lookups**: Always check `if key not in target_dict` before comparing `target_dict[key]` against a threshold.
4. **Update & Terminate**: Append matching items via `.append()` and ensure `while` counters increment on every pass.
