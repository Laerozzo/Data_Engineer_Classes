## Chapter 1: Introduction and Flat Files

### 1.1 Reading Plain Text Files & Context Managers (`with`)
* **`open(filename, mode)`**: Opens a connection to a file (`mode='r'` for read-only, `mode='w'` for write).
* **Context Manager (`with open(...) as file:`)**: Best practice because it **automatically closes** the file once the block finishes executing, preventing file locks or accidental modifications.
* **Reading Methods**:
  * `file.read()`: Reads the entire file content into a single string.
  * `file.readline()`: Reads one line at a time sequentially.

### 1.2 Importing Flat Files with `NumPy`
* Flat files are basic text tables (like CSV or tab-delimited files) containing records without structured relationships.
* If all data is numeric (e.g., image matrices like MNIST), `np.loadtxt()` loads data directly into a fast NumPy `ndarray`:
  * `delimiter`: Character separating values (`','` for CSV, `'\t'` for tab-delimited).
  * `skiprows`: Number of leading rows to skip (e.g., `skiprows=1` to skip headers).
  * `usecols`: List of column indices to import (e.g., `usecols=[0, 2]`).
  * `dtype`: Data type of the resulting array (`float` by default, or `str`).
* For mixed data types in NumPy, `np.genfromtxt()` or `np.recfromcsv()` can handle structured arrays, which motivates using `pandas`.

### 1.3 Importing Flat Files with `pandas`
* `pd.read_csv()` imports flat files with mixed data types into a two-dimensional `DataFrame`:
  * `nrows`: Limits import to the first $n$ rows.
  * `header=None`: Indicates the file has no header row.
  * `sep`: Specifies custom delimiter (`','`, `'\t'`, `';'`).
  * `comment='#'`: Ignores characters following `#` on any line.
  * `na_values=['Nothing']`: Recognizes custom strings as `NaN`/missing values.
* Use `df.to_numpy()` (or `.values`) to convert a DataFrame into a NumPy array.

```python
# Using NumPy to import flat files (digits.csv & digits_header.txt)
import numpy as np
import matplotlib.pyplot as plt

file = 'datasets/digits.csv'
digits = np.loadtxt(file, delimiter=',')
print("Datatype of digits:", type(digits))

# Select and reshape a row into a 28x28 image
im = digits[21, 1:]
im_sq = np.reshape(im, (28, 28))

plt.imshow(im_sq, cmap='Greys', interpolation='nearest')
plt.title("Reshaped 28x28 Digit Array")
plt.show()

# Customizing NumPy import with skiprows and usecols
data = np.loadtxt('datasets/digits_header.txt', delimiter='\t', skiprows=1, usecols=[0, 2])
print("Subset of digits_header.txt (cols 0 and 2):\n", data[:5])
```

```python
# Importing seaslug.txt as strings and floats
file = 'datasets/seaslug.txt'

# Import file as strings
data = np.loadtxt(file, delimiter='\t', dtype=str)
print("Header row:", data[0])

# Import file as floats and skip the first row
data_float = np.loadtxt(file, delimiter='\t', dtype=float, skiprows=1)
print("10th element:", data_float[9])

plt.scatter(data_float[:, 0], data_float[:, 1])
plt.xlabel('time (min.)')
plt.ylabel('percentage of larvae')
plt.title('Seaslug Larvae Data')
plt.show()
```

```python
# Importing flat files using pandas (titanic.csv, digits.csv, titanic_corrupt.txt)
import pandas as pd

# 1. Read standard CSV
df = pd.read_csv('datasets/titanic.csv')
print(df.head(3))

# 2. Read first 5 rows with no header and convert to NumPy array
data = pd.read_csv('datasets/digits.csv', nrows=5, header=None)
data_array = data.to_numpy()
print("Converted NumPy array type:", type(data_array))

# 3. Handle corrupt tab-delimited file with comments and custom missing values
data_corrupt = pd.read_csv('datasets/titanic_corrupt.txt', sep='\t', comment='#', na_values=['Nothing'])
print("\nCleaned titanic_corrupt.txt:")
print(data_corrupt.head())
```

### Chapter 1: Key Takeaways & Actionable Step-by-Step Process

#### Key Takeaways
* Always use the `with open(...) as file:` context manager for plain text files to guarantee automatic file closure.
* Use `np.loadtxt()` when importing purely numerical matrices (like image pixel arrays) and `pd.read_csv()` when working with tabular datasets containing mixed types, headers, and missing values.
* `pd.read_csv()` can clean corrupt files on import using `sep`, `comment`, and `na_values`.

#### Step-by-Step Process to Act On
1. **Inspect File Format**: Check if the file is plain text, a pure numerical matrix, or a mixed-type table, and identify the delimiter (`','`, `'\t'`, `';'`).
2. **Choose the Loader**: Use `with open()` for unstructured text, `np.loadtxt()` for numeric arrays, or `pd.read_csv()` for tabular datasets.
3. **Configure Import Parameters**: Pass `skiprows`, `usecols`, `comment`, and `na_values` directly to the reader function to handle headers and corrupt tokens upfront.
4. **Verify Import**: Print `.head()`, `.shape`, or `type()` immediately after loading.

## Chapter 2: Importing Data from Other File Types

### 2.1 Pickled Files (`.pkl`) & `os` Directory Listing
* **Pickle (`import pickle`)**: Serializes native Python objects (dicts, lists, models) into binary files. Open with `mode='rb'` (read binary) and load via `pickle.load(file)`.
* **Directory Inspection**: `os.listdir(os.getcwd())` lists all files in the current directory.

### 2.2 Excel Spreadsheets (`.xlsx`)
* Load multi-sheet workbooks using `xls = pd.ExcelFile('file.xlsx')`.
* Inspect sheet names via the `xls.sheet_names` attribute.
* Parse individual sheets into DataFrames using `xls.parse(sheet_name_or_index, skiprows=..., names=..., usecols=...)`.

### 2.3 SAS (`.sas7bdat`) and Stata (`.dta`) Files
* **SAS**: Used in business analytics and biostatistics. Imported via `from sas7bdat import SAS7BDAT` (or `pd.read_sas('file.sas7bdat')`).
* **Stata**: Used in academic social sciences/economics. Imported directly via `pd.read_stata('file.dta')` without needing a `with` block.

### 2.4 HDF5 (`.hdf5`) and MATLAB (`.mat`) Files
* **HDF5 (`import h5py`)**: Hierarchical Data Format version 5 for storing massive numerical datasets (e.g., LIGO gravitational wave data). Loaded with `h5py.File(filename, 'r')` and traversed like nested Python dictionaries (`data['group']['dataset']`).
* **MATLAB (`import scipy.io`)**: Matrix Laboratory `.mat` workspaces loaded into a Python dictionary via `scipy.io.loadmat(filename)` where keys correspond to MATLAB variable names and values are NumPy arrays.

```python
# Loading a pickled file and inspecting directory contents
import pickle
import os

print("Files in datasets directory:", os.listdir('datasets'))

with open('datasets/data.pkl', 'rb') as file:
    d = pickle.load(file)

print("Loaded pickle content:", d)
print("Datatype of d:", type(d))
```

```python
# Loading Excel spreadsheets and customizing .parse()
import pandas as pd

file = 'datasets/battledeath.xlsx'
xls = pd.ExcelFile(file)

# Print sheet names
print("Sheet names:", xls.sheet_names)

# Load a sheet by name and by index, customizing headers and skipped rows
df1 = xls.parse(0, skiprows=[1], names=['Country', 'AAM due to war (2002)'])
print("\nParsed Sheet 0:")
print(df1.head())

df2 = xls.parse(1, usecols=[0], skiprows=[1], names=['Country'])
print("\nParsed Sheet 1 (first column only):")
print(df2.head())
```

```python
# Importing Stata (.dta), HDF5 (.hdf5), and MATLAB (.mat) files
import pandas as pd
import h5py
import scipy.io
import numpy as np

# 1. Stata file
df_stata = pd.read_stata('datasets/disarea.dta')
print("Stata DataFrame head:")
print(df_stata.head(3))

# 2. HDF5 file
h5_data = h5py.File('datasets/dataset.hdf5', 'r')
print("\nHDF5 top-level keys:", list(h5_data.keys()))
print("HDF5 Strain sample:", h5_data['strain']['Strain'][:5])
h5_data.close()

# 3. MATLAB file
mat = scipy.io.loadmat('datasets/albeck_gene_expression.mat')
print("\nMATLAB dictionary keys:", mat.keys())
print("Type of CYratioCyt:", type(mat['CYratioCyt']))
print("Shape of CYratioCyt:", np.shape(mat['CYratioCyt']))
```

### 3.1 Relational Databases & Creating a Database Engine
* **Relational Model (Codd's 12 Rules)**: Organizes data into tables (relations) linked by unique **Primary Keys** and **Foreign Keys**.
* **`SQLAlchemy` / `sqlite3`**: Connects Python to SQL databases (`SQLite`, `PostgreSQL`, `MySQL`).
* **Workflow for Querying Databases**:
  1. Import database driver (`sqlite3` or `from sqlalchemy import create_engine, inspect, text`).
  2. Create a connection/engine (`conn = sqlite3.connect('Chinook.sqlite')`).
  3. Open a context manager (`with conn:`) and execute SQL queries (`SELECT * FROM Table WHERE ... ORDER BY ... INNER JOIN ...`).
  4. Fetch results (`cursor.fetchall()` or `cursor.fetchmany(size)`) and wrap in `pd.DataFrame()`, **or** use the one-line shortcut `pd.read_sql_query(query, conn)`.

```python
# Querying the mock Chinook.sqlite relational database using sqlite3 and pandas
import sqlite3
import pandas as pd

# Connect to Chinook.sqlite
conn = sqlite3.connect('datasets/Chinook.sqlite')

# Inspect table names in the SQLite database
tables = pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table';", conn)
print("Tables in Chinook.sqlite:")
print(tables)

# Execute SQL query with WHERE and ORDER BY using pd.read_sql_query()
df_employees = pd.read_sql_query(
    "SELECT EmployeeId, FirstName, LastName, Title, BirthDate FROM Employee WHERE EmployeeId >= 3 ORDER BY BirthDate;",
    conn
)
print("\nFiltered & Ordered Employees:")
print(df_employees)

# Execute INNER JOIN across Album and Artist tables
join_query = """
SELECT Album.Title AS AlbumTitle, Artist.Name AS ArtistName
FROM Album
INNER JOIN Artist ON Album.ArtistId = Artist.ArtistId;
"""
df_joined = pd.read_sql_query(join_query, conn)
print("\nJoined Album & Artist Records:")
print(df_joined.head())

conn.close()
```

### Chapter 3: Key Takeaways & Actionable Step-by-Step Process

#### Key Takeaways
* Relational databases store normalized data across tables linked by keys (`ArtistId`, `EmployeeId`).
* While manual cursors (`fetchall()`, `fetchmany()`) offer fine-grained control, `pd.read_sql_query(sql, con)` executes SQL queries and returns a labeled `DataFrame` in a single line.
* Pushing filtering (`WHERE`), sorting (`ORDER BY`), and table joining (`INNER JOIN`) into SQL before loading into Python saves memory and improves performance.

#### Step-by-Step Process to Act On
1. **Establish Connection**: Create a database engine (`create_engine()`) or connection (`sqlite3.connect()`).
2. **Inspect Schema**: Query table names (`sqlite_master` or SQLAlchemy `inspect(engine).get_table_names()`) to understand available relations.
3. **Write Targeted SQL**: Filter rows with `WHERE` and combine related tables using `INNER JOIN ... ON` directly in your SQL string.
4. **Load into DataFrame**: Pass the query and connection to `pd.read_sql_query()` and close the connection when finished.
