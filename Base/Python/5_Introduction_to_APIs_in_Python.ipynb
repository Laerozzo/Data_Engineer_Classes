1. urllib.request.urlopen(url)
This function opens a network connection to the specified URL and sends an HTTP request.
• What it does: It acts like typing a address into your browser and pressing Enter. It connects to the remote server, sends a GET request by default, and waits for the server to reply.
• What it returns: An HTTPResponse object. This object acts like a streaming pipe (or file) connected directly to the server's response data.


2. response.read()
This method downloads the raw content from the open network stream.
• What it does: It reads the data flowing through the response pipe. Because network data travels over the wire as raw electricity or light pulses, Python reads this as raw binary.
• What it returns: A bytes object (e.g., b'{"id": 1}'). It pulls the entire payload into your computer's memory.

3. .decode('utf-8')
This method converts raw bytes into a Python string.
• What it does: Raw binary (bytes) cannot be read like normal text. .decode('utf-8') translates those 1s and 0s into readable text characters based on the UTF-8 character encoding standard.
• What it returns: A standard Python str object.
• Note: In newer versions of Python (3.6+), json.loads() can actually handle raw bytes directly, meaning you can often skip .decode() entirely!

4. json.loads(string_or_bytes)
This function parses a JSON text string and converts it into a Python data structure.
• What it does: "loads" stands for "load string". It takes a raw string (or bytes) formatted as JSON (JavaScript Object Notation) and maps its syntax to Python counterparts (e.g., turning JSON {} into a Python dict, and [] into a list).
• What it returns: A native Python object—most commonly a dict or a list.

When you call requests.get(url), the library executes the following four-step pipeline automatically:

1. Session and Connection Pooling
• What it does: Instead of starting from scratch, requests spins up an internal connection manager.
• The Benefit: It automatically reuses TCP connections (Keep-Alive) if you make multiple calls to the same server, speeding up your program.

2. Request Construction & Sending
• What it does: It packages your parameters, cookies, and HTTP headers into a properly formatted request payload.
• The Benefit: It automatically identifies itself to the server using a default user-agent and handles complex redirects (like changing http to https) silently.

3. Automatic Response Management
• What it does: It intercepts the raw binary data stream coming back from the server, downloads it into your system's memory, and then closes the connection instantly for you.
• The Benefit: You never have to manually call .close(). There is no risk of leaking network sockets or causing memory bugs.

4. Intelligently Parsing the Response
• What it does: It reads the server's Content-Type header (such as text/html or application/json) and automatically figures out the correct text encoding (like UTF-8).
• The Benefit: It prepares the data so you can access it immediately as text, raw bytes, or a Python dictionary.

### The Anathomy of an API URL

An API URL (more formally called a URI or Endpoint) is a structured address designed to tell a server exactly what data you want and what actions you want to take.
Here is the anatomy of a complete API URL, broken down by its structural components:

  https://   api.example.com   /v2   /users   /123  ?status=active&sort=desc
  └──┬───┘  └──────┬───────┘  └┬─┘  └──┬──┘  └─┬─┘  └──────────┬───────────┘
   Protocol     Host/Domain  Version Resource  ID     Query Parameters

1. Protocol (Scheme)
• What it is: https:// or http://
• What it does: Specifies the rules used to transfer data between your code and the server. HTTPS is the industry standard for APIs because it encrypts the data during transit.

2. Host / Domain
• What it is: ://example.com or ://example.com
• What it does: Points to the location of the server hosting the API. Many organizations use a dedicated api. subdomain to separate their data services from their main website traffic.

3. API Version
• What it is: /v1, /v2, or /v3
• What it does: Prevents breaking changes. If a company updates how their API works, they deploy a new version (like v2) so that older applications using v1 don't suddenly crash.

4. Resource (Path)
• What it is: /users, /products, /orders
• What it does: Represents the noun or category of data you are trying to access. In RESTful API design, paths are almost always named as plural nouns.

5. Identifier (ID)
• What it is: /123 or /usr_9j2f8
• What it does: Narrows down the scope from a broad collection to a single, specific record. For example, /users grabs all users, while /users/123 targets only the user with ID 123.

6. Query Parameters
• What it is: ?status=active&sort=desc
• What it does: Filters, sorts, or limits the data returned by the server.
	• It always begins with a question mark (?).
	• Parameters are written as key-value pairs (key=value).
	• Multiple parameters are separated by ampersands (&).

### Adding query parameters with requests:

- We can either append the query parameter to the url:
https://example.com?status=active&sort=desc&limit=5

- Or, we can use params argument inside requests.get(url, params=dict) to add query parameters:

query_parameters = {
    'status': 'active',
    'sort': 'desc',
    'limit': 5
}

response = requests.get(url, params=query_parameters)


### Major Benefits of Using params
1. Automatic URL Encoding (Handling Spaces & Special Characters)

URLs cannot contain spaces or special characters (like #, &, or ?). If your parameter contains these, they must be converted into standard web-safe codes (percent-encoding). The params argument does this silently.

"" 
payload = {'search': 'shoes & boots'}
response = requests.get('https://httpbin.org', params=payload)

print(response.url)
#Output: https://httpbin.org?search=shoes+%26+boots
#Notice how the space became '+' and '&' became '%26' automatically!


2. Clean Multi-Value Parameters (Lists)
If you need to pass multiple values for the same key (e.g., filtering for multiple statuses at once), you can pass a list as the dictionary value.

"" 
payload = {'tags': ['python', 'networking', 'api']}
response = requests.get('https://httpbin.org', params=payload)

print(response.url)
# Output: https://httpbin.org?tags=python&tags=networking&tags=api


3. Automatic Stripping of None Values
If your program conditionally adds parameters, any key set to None will be completely ignored and stripped out of the final URL, preventing syntax errors on the backend server.

"" 
payload = {
    'page': 1,
    'search_query': None  # This will be cleanly ignored
}
response = requests.get('https://httpbin.org', params=payload)

### HTTP Verbs:

GET // POST // PUT // DELETE


1. GET (Read)
Used exclusively to fetch data from a server. It should never modify data on the backend. Because data is passed via the URL, it can be easily bookmarked or cached.

#Fetches a list of all products
response = requests.get('https://example.com')


2. POST (Create)
Used to submit new data to the server to create a brand new record (like registering a user or submitting a form). Every time you run a POST request, a new entry is generated.

#Creates a brand new product in the system
new_product = {"name": "Wireless Mouse", "price": 29.99}
response = requests.post('https://example.com', json=new_product)


3. PUT (Update/Replace)
Used to update an existing record by overwriting it completely. You target a specific item by its ID. If the item doesn't exist, some APIs will use PUT to create it. If you miss a field in your update payload, that missing field might be wiped out on the server.

#Replaces product #45 with this updated dataset
updated_product = {"name": "Gaming Wireless Mouse", "price": 45.00}
response = requests.put('https://example.com', json=updated_product)


4. DELETE (Delete)
Used to remove a specific record from the server using its unique identifier.

#Permanently deletes product #45
response = requests.delete('https://example.com')

### Headers and status codes


The Most Common API Status Codes

🟢 2xx Success (Everything is fine)
• 200 OK: The standard success code. Your GET request worked, and the data is in the response.
• 201 Created: Typically returned after a successful POST request. The new resource has been created.
• 204 No Content: The request succeeded, but there is no data to send back (common for successful DELETE or PUT requests).

🟡 3xx Redirection (The resource moved)
• 301 Moved Permanently: The URL you requested has been assigned a new permanent URL.
• 304 Not Modified: Used for caching. The data hasn't changed since your last visit, so you can reuse the data you already have.

🔴 4xx Client Error (Fix your Python code)
• 400 Bad Request: The server cannot understand your request (e.g., your JSON payload has a typo or syntax error).
• 401 Unauthorized: You lack valid authentication credentials. You either forgot to send an API key or your token is expired.
• 403 Forbidden: The server understands who you are, but you do not have permission to access this specific resource.
• 404 Not Found: The resource or endpoint does not exist. Check your URL spelling.
• 429 Too Many Requests: You hit the API's rate limit. You are sending requests too fast.

💥 5xx Server Error (The website is broken)
• 500 Internal Server Error: A generic error message given when an unexpected condition was encountered on the server's backend.
• 502 Bad Gateway: The server acting as a gateway received an invalid response from the upstream server.
• 503 Service Unavailable: The server is temporarily overloaded or down for maintenance.

### Headers

HTTP headers work like the shipping label on a package.
When you make an API request, the headers provide metadata (extra information) about the payload, the client making the request, and the expected response format. This metadata is separate from the actual content (body) of your request.


### The Most Common Header Categories

1. Authentication (Who are you?)
Most APIs require you to prove your identity before they will share data.
• Authorization: Used to pass API keys, tokens, or credentials.
	• Example: Authorization: Bearer your_api_token_here

2. Content Negotiation (What data format do you want?)
This tells the server what kind of data formats your code can understand or is currently sending.
• Accept: Tells the server what format you want back.
	• Example: Accept: application/json (Gimme JSON, not HTML)
• Content-Type: Tells the server what format you are sending in a POST or PUT request.
	• Example: Content-Type: application/json (Heads up, I'm sending JSON)

3. Identification (What app or device are you?)
• User-Agent: Identifies the software or library making the request. Some servers block requests that look like generic scripts to prevent bot abuse.
	• Example: User-Agent: Mozilla/5.0 or User-Agent: MyApp/1.0


### How to Use Headers in Python requests

Reading headers from a response:

response = requests.get('API URL')
print(response.headers['content-type'])

Just like the params argument, you pass custom headers into requests.get() or requests.post() using a Python dictionary via the headers parameter.

import requests

url = 'https://example.com'

#Define your metadata/credentials in a dictionary
custom_headers = {
    'Authorization': 'Bearer my-secret-api-key-123',
    'Accept': 'application/json',
    'User-Agent': 'MyCustomPythonApp/1.0'
}

#Pass it to the headers argument
response = requests.get(url, headers=custom_headers)

#You can also look at the headers the SERVER sent back to you!
server_headers = response.headers
print(server_headers.get('Content-Type')) 
#Output might be: application/json; charset=utf-8

Respecting API rate limits

- Python:

import requests
import time

longestTrackLength = 0
longestTrackTitle = ""
headers = {'Authorization': 'Bearer 8apDFHaNJMxy8Kt818aa6b4a0ed0514b5d3'}
page_number = 1

while True:
    params = {'page': page_number, 'per_page': 500}
    response = requests.get('http://localhost:3000/tracks', params=params, headers=headers)
    response.raise_for_status()
    response_data = response.json()
    
    print(f'Fetching tracks page {page_number}')

    if len(response_data['results']) == 0:
        break

    for track in response_data['results']:
        if(track['Length'] > longestTrackLength):
            longestTrackLength = track['Length']
            longestTrackTitle = track['Name']

    page_number = page_number + 1
    
    # 3 second pause at the end of while loop, without this, we would get the 429 Client Error - indicating we are sending too many requests.
    time.sleep(3)

print('The longest track in my music library is: ' + longestTrackTitle)
