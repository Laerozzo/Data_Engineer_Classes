Importing entire text files

Reading a text file:

    filename = 'huck_fin.txt'
    file = open(filename, mode='r') #'r' is to read //  file = open(filename, mode='w') #'w' is to write
    text = file.read()
    file.close()
    print(text)

Context manager 'WITH'

    with open('huck_finn.txt','r') as file: 
        print(file.read())

using the context manager guarantees that you are closing the file after opening it, to avoid any unwanted edits to the file.

    #Open a file as read-only and bind it to file
        with open('moby_dick.txt', mode='r') as file:
  	    # Print it
