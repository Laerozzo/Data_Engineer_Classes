```python
full_name = "Alan Turing"

# Define the generate_email function
def generate_email(full_name):
    name_parts = full_name.split()
    email = name_parts[0].lower() + '.' + name_parts[1].lower() + '@techcompany.com'
    
    # Return the email address
    return email

# Call the function on the full_name string
print(generate_email(full_name))
```

```python
test_durations = [245.50, 189.99, 312.75, 156.20, 428.90, 201.35, 167.80]

 # Complete the function
def test_report(durations):
    num_tests = len(durations)
    
    # Calculate total test time
    total_time = sum(durations)
    
    print("=== Test Report ===")
    print("Total Tests: ", num_tests)
    print("Total Execution Time (s): ", total_time)

# Generate the report for recent test runs
test_report(test_durations)
```

```python
import string
user_password = "1233124as?f"

def validate_password(password):
    # Check if password is at least 8 characters long
    if len(password) >= 8:
        # Check if password contains a special character
        for char in password:
            if char in string.punctuation:
                return True
    return False

# Call the function and store the result
is_valid = validate_password(user_password)
print("Is the password valid? ", is_valid)
```

Arbitrary arguments allow functions to accept any number of arguments

    #function code remains the same

* : converts arguments to a single iterable (tuple)

now, average can be called with multiple arguments:

Arbitrary keyword arguments: **kwargs

keyword = value

def average(**kwargs):
    average_avlue = sum(kwargs.values()) / len(kwargs.values())
    rounded_average = round(average_value, 2)
    return rounded_average

with kwargs, we must give each argument a name 
average (a=2, b=15, c=18, d=25) < returns average of 2,15,18,25 (or a+b+c+d)>
the function combines the single names and values into a dictionary {}

can also be called like this:
average(**{}) < ** this forces pyhthon to take each dictionary key as a kwarg and each value as the assigned value>

we can combine multiple args or kwargs, to "join" multiple arguments into a single one and return the value 


```python
# Define a function called concat
def concat(*arg):
  """Concatenates multiple string arguments with spaces between them."""

  result = ""

  # Iterate over the Python args tuple
  for arg in arg:
    result += " " + arg
  return result

# Call the function
print(concat("Python", "is", "great!"))
```

```python
# Define a function called concat
def concat(**kwargs):
  """Concatenates keyword arguments into a single string with spaces."""
  
  result = ""
  
  # Iterate over the Python kwargs
  for kwarg in kwargs.values():
    result += " " + kwarg
  return result

# Call the function
print(concat(start="Python",middle="is",end="great!"))
```

## Chapter 3 - Lambda Functions

Lambda function is a quick way for defining an anonymous function that doesn't need a name or to be stored in a variable.
Syntax:
    lambda x,y,z: expression

no return statement is required 
even though not needed, it can still be stored as a variable

Example, lambda function for calculating the Average:

print(lambda x: sum(x) / len(x))

Works the same as

def average(x):
    return sum(x)/len(x)
print(average)

to use a lambda function, we place it inside parenthesis, then, we open parenthesis again to add the arguments

    (lambda x: sum(x)/len(x))([3,6,9])

    print((lambda x: sum(x)/len(x))([3,6,9]))

We can store lambda functions as a variable if we intend to use it multiple times:

    average = lambda x: sum(x)/len(x)

#calling the average function
    print(average([3,6,9]))

We can use multiple parameters:

    lambda: x,y: x ** y

Lambda functions with iterables:

the map() function, applies a function (lambda in this case) to all elements in an iterable (list, dictionary, tuple)

    names = []
    #Apply a lambda function inside map()
    capitalize = map(lambda x: x.capitalize(), names)

map() syntax:
    map(function, iterable1, iterable2, ...)

map() returns a map object, not the iterable. In order to return an iterable, we must call the iterable function and create a new list/dictionary/tuple using the map object generated:

    names = [""]
    capitalize = map(lambda x: x.capitalize(), names)
    print(capitalize)
    <returned map object>

    print(list(capitalize))
    <returns a list with names capitalized>




```python
file_size = 2500
extra_space = 0.15

# Define a lambda function
calculate_total = lambda x: x * (1 + extra_space)

# Call the lambda function
print(calculate_total(file_size))
```
