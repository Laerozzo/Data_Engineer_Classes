## Chapter 1: The Python Ecosystem

### 1.1 Built-in Functions
* Python ships with standard built-in functions that operate on iterables without requiring imports:
  * `min(iterable)` / `max(iterable)`: Returns the smallest or largest element.
  * `sum(iterable)`: Calculates the total sum of numeric elements.
  * `len(iterable)`: Returns the count of elements in a collection or characters in a string.
  * `round(number, ndigits)`: Rounds a floating-point number to `ndigits` decimal places.
  * `sorted(iterable)`: Returns a new sorted list in ascending order by default.

### 1.2 Modules
* **Overview**: Python scripts (`.py` files) containing reusable functions, classes, and attributes.
* **`os` Module**: Interacts with the operating system (`import os`):
  * `os.getcwd()`: Function that returns the current working directory path.
  * `os.chdir(path)`: Function that changes the current working directory.
  * `os.environ`: Attribute (not called with `()`) returning a mapping of environment variables.
* **`string` Module**: Provides useful character constants (`import string`) for input validation:
  * `string.ascii_lowercase`, `string.digits`, `string.punctuation`.

### 1.3 Packages and `pandas`
* **Packages (Libraries)**: Collections of modules downloadable from PyPI via terminal (`python3 -m pip install pandas`).
* **`pandas` DataFrames**: Tabular data structures created from dictionaries (`pd.DataFrame(dict_data)`) or external CSV files (`pd.read_csv("file.csv")`). Inspect using `.head()` and `.info()`.

```python
recent_orders = [15.99, 28.50, 42.75, 18.99, 55.00, 31.25, 22.99, 67.50]

# Find the smallest and largest order amounts, and total revenue
smallest_order = min(recent_orders)
largest_order = max(recent_orders)
total_revenue = sum(recent_orders)

print("Smallest order:", smallest_order)
print("Largest order:", largest_order)
print("Total revenue:", round(total_revenue, 2))
```

```python
delivery_times = [19, 25, 35, 40, 28, 32, 29, 31]

# Find the average delivery time and round to two decimal places
average_time = sum(delivery_times) / len(delivery_times)
average_rounded = round(average_time, 2)

print("Average delivery time:", average_rounded)
```

```python
restaurants = ["Sushi Central", "Burger Hub", "Taco Town", "Pizza Palace"]
cooking_times = [30, 25, 35, 40, 28, 32, 29, 31, 12, 55]

# Sort restaurant names alphabetically and cooking times ascending
restaurants_sorted = sorted(restaurants)
cooking_times_sorted = sorted(cooking_times)

print("Restaurants (A–Z):", restaurants_sorted)
print("Cooking times (ascending):", cooking_times_sorted)
```

## Chapter 2: Creating and Defining Custom Functions

### 2.1 Defining Functions (`def` and `return`)
* Use the `def` keyword followed by the function name, parameters in parentheses `()`, and a colon `:`.
* `return` outputs the result back to the caller and exits the function immediately.

### 2.2 Positional, Keyword, and Default Arguments
* **Positional Arguments**: Passed in exact order of definition.
* **Keyword Arguments (`param=value`)**: Passed explicitly by parameter name; order does not matter.
* **Default Arguments**: Defined in the function signature (e.g., `def clean_text(text, lower=True):`), providing a fallback value if the caller omits that argument.

### 2.3 Docstrings (`__doc__`)
* Placed on the first line inside a function using triple quotes (`"""..."""`).
* Accessible via `func.__doc__` (dunder doc attribute) or `help(func)`.
* Multi-line docstrings should document **Summary**, **Args**, and **Returns**.

### 2.4 Arbitrary Arguments (`*args` and `**kwargs`)
* **`*args`**: Collects any number of positional arguments into a single **tuple**.
* **`**kwargs`**: Collects any number of keyword arguments (`key=value`) into a single **dictionary**.

```python
full_name = "Alan Turing"

# Define the generate_email function
def generate_email(full_name):
    name_parts = full_name.split()
    email = name_parts[0].lower() + '.' + name_parts[1].lower() + '@techcompany.com'
    return email

print(generate_email(full_name))
```

```python
test_durations = [245.50, 189.99, 312.75, 156.20, 428.90, 201.35, 167.80]

def test_report(durations):
    num_tests = len(durations)
    total_time = sum(durations)
    print("=== Test Report ===")
    print("Total Tests:", num_tests)
    print("Total Execution Time (s):", round(total_time, 2))

test_report(test_durations)
```

```python
import string
user_password = "1233124as?f"

def validate_password(password):
    if len(password) >= 8:
        for char in password:
            if char in string.punctuation:
                return True
    return False

is_valid = validate_password(user_password)
print("Is the password valid?", is_valid)
```

```python
def clean_text(text, lower=True):
    """
    Clean text by swapping spaces to underscores and converting to lowercase.
    
    Args:
        text (str): A string to be cleaned.
        lower (bool): Whether to convert the text to lowercase. Defaults to True.
    
    Returns:
        str: Cleaned string.
    """
    cleaned = text.replace(' ', '_')
    if not lower:
        return cleaned
    return cleaned.lower()

print(clean_text('Abracada  Bra'))
print(clean_text('Abra ca DabrA', lower=False))
print(clean_text.__doc__)
```

```python
original_price = 899.99

def calculate_discount(price, discount_percent=15, round_result=True):
    """Calculate the discounted price of a product."""
    discounted_price = price - (price * (discount_percent / 100))
    if round_result:
        return round(discounted_price, 2)
    return discounted_price

final_price = calculate_discount(price=original_price, discount_percent=25, round_result=False)
print("Final price:", final_price)
print("Docstring:", calculate_discount.__doc__)
```

```python
# Arbitrary positional (*args) and keyword (**kwargs) arguments
def concat_args(*args):
    """Concatenates multiple string arguments with spaces between them."""
    result = ""
    for arg in args:
        result += " " + arg
    return result.strip()

def concat_kwargs(**kwargs):
    """Concatenates keyword arguments into a single string with spaces."""
    result = ""
    for kwarg in kwargs.values():
        result += " " + kwarg
    return result.strip()

print(concat_args("Python", "is", "great!"))
print(concat_kwargs(start="Python", middle="is", end="great!"))
```

## Chapter 3: Lambda Functions and Error Handling

### 3.1 Lambda Functions
* **Overview**: Anonymous, inline functions defined with the `lambda` keyword (`lambda parameters: expression`).
* Automatically returns the evaluated expression without an explicit `return` keyword.
* Ideal for short, throwaway transformations inside higher-order functions like `map()`.
* **`map(function, iterable)`**: Applies `function` to every item in `iterable` and returns a lazy `map` object (wrap in `list()` to materialize).

### 3.2 Introduction to Errors & Tracebacks
* **Common Built-in Exceptions**:
  * `SyntaxError`: Code violates Python grammar rules (e.g., missing comma or colon).
  * `NameError`: Referencing a variable or function name that has not been defined.
  * `TypeError`: Applying an operation or function to an incompatible data type (e.g., `'5' + 5`).
  * `ValueError`: Passing an argument of the right type but inappropriate value (e.g., `int('abc')` or negative value where positive is required).
* **Tracebacks**: Diagnostic error reports showing the call stack and exact line where an exception was raised (including internal package source files).

### 3.3 Error Handling (`try` / `except` and `raise`)
* **Context Added**: Two main ways to handle errors in custom functions:
  1. **`try` / `except` Block**: Attempts to run code inside `try:`; if an exception occurs, execution jumps gracefully to the `except:` block instead of crashing.
  2. **`raise` Keyword**: Explicitly triggers a specific exception (e.g., `raise TypeError("Input must be a list")`) with a clear message when input validation fails.

```python
file_size = 2500
extra_space = 0.15

# Store a lambda function in a variable and call it
calculate_total = lambda x: x * (1 + extra_space)
print(calculate_total(file_size))

# Call a lambda function inline in one line
print((lambda x: x * (1 + extra_space))(file_size))
```

```python
colleagues = ["Sarah Martinez", "Michael Chen", "Emily Brown"]

# Apply a lambda function to each colleague's name using map()
cleaned = map(lambda x: x.replace(" ", "_").lower(), colleagues)

# Convert map object to a list
cleaned_list = list(cleaned)
print(cleaned_list)
```

```python
# Complete Error Handling Examples: try-except vs. raise
def snake_case_try(text):
    """Attempt to convert text to snake_case using try-except."""
    try:
        clean_text = text.replace(" ", "_").lower()
        return clean_text
    except AttributeError:
        print("Error: The input provided must be a string data type.")

def snake_case_raise(text):
    """Convert text to snake_case, raising TypeError if input is not a string."""
    if not isinstance(text, str):
        raise TypeError("The 'text' argument must be of type str, got " + type(text).__name__)
    return text.replace(" ", "_").lower()

# Test valid and invalid calls
print("Valid call:", snake_case_try("User Name 01"))
snake_case_try(12345)

try:
    snake_case_raise(12345)
except TypeError as err:
    print("Caught raised exception:", err)
```

### Chapter 3: Key Takeaways & Actionable Step-by-Step Process

#### Key Takeaways
* `lambda` functions combined with `map()` provide concise, functional-style transformations over iterables.
* Understanding tracebacks (`SyntaxError`, `NameError`, `TypeError`, `ValueError`) speeds up debugging both local scripts and third-party package calls.
* Defensive programming uses `try`/`except` to catch runtime failures and `raise` (with `isinstance()`) to enforce strict input contracts.

#### Step-by-Step Process to Act On
1. **Write Inline Transforms**: Use `list(map(lambda x: ..., iterable))` for single-expression element transformations.
2. **Read Tracebacks Bottom-Up**: Check the final line of a traceback first for the exception type and message, then look up to find the offending line in your code.
3. **Validate Function Inputs**: Use `if not isinstance(arg, expected_type): raise TypeError(...)` at the top of custom functions.
4. **Wrap Risky Operations**: Place external operations or type conversions inside `try:` blocks and catch specific exceptions in `except:`.
