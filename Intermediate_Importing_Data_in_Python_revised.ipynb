## Chapter 1: Importing Data from the Internet

### 1.1 Automating File Downloads (`urllib` & `pandas`)
* **`urllib.request.urlretrieve(url, filename)`**: Downloads a remote file located at `url` and saves it locally under `filename`.
* **Direct `pandas` Web Import**: Both `pd.read_csv(url, sep=';')` and `pd.read_excel(url, sheet_name=None)` accept remote URLs directly without saving a local copy first. Passing `sheet_name=None` to `pd.read_excel()` loads **all sheets** into an ordered dictionary where keys are sheet names and values are DataFrames.

### 1.2 HTTP GET Requests (`urllib` vs. `requests`)
* **Using `urllib.request`**:
  1. Package request: `request = Request(url)`
  2. Send & catch response: `response = urlopen(request)`
  3. Extract raw HTML bytes: `html = response.read()`
  4. Close connection: `response.close()`
* **Using `requests` (Higher-level & Cleaner)**:
  1. Send & catch response in one call: `r = requests.get(url)`
  2. Extract HTML as a decoded string via attribute: `text = r.text` (no manual `.close()` needed).

### 1.3 Web Scraping with `BeautifulSoup`
* Raw HTML is a mix of structured tags and unstructured text. `BeautifulSoup` (`from bs4 import BeautifulSoup`) parses HTML into a navigable tree object (`soup = BeautifulSoup(html_doc, 'html.parser')`):
  * `soup.prettify()`: Formats HTML with clean indentation.
  * `soup.title`: Attribute returning the `<title>` tag.
  * `soup.get_text()` (or `soup.text`): Extracts all human-readable text without HTML tags.
  * `soup.find_all('a')`: Finds all occurrences of a specific tag (e.g., `<a>` hyperlink tags) so you can iterate and extract attributes via `link.get('href')`.

```python
# Importing flat and Excel files from local mock files / web URLs using pandas
import matplotlib.pyplot as plt
import pandas as pd

# Read winequality-red.csv (semicolon-separated)
df = pd.read_csv('datasets/winequality-red.csv', sep=';')
print(df.head())

# Plot histogram of the first column
df.iloc[:, 0].hist()
plt.xlabel('fixed acidity (g(tartaric acid)/dm$^3$)')
plt.ylabel('count')
plt.title('Distribution of Fixed Acidity')
plt.show()

# Read all sheets of latitude.xlsx using sheet_name=None
xls = pd.read_excel('datasets/latitude.xlsx', sheet_name=None)
print("\nSheet names (dictionary keys):", xls.keys())
print("\nFirst sheet ('1700') head:")
print(xls['1700'].head())
```

```python
# Performing HTTP requests and parsing HTML with BeautifulSoup
from bs4 import BeautifulSoup

# Using a sample HTML string representing Guido van Rossum's homepage for reproducible offline execution
sample_guido_html = """
<!DOCTYPE html>
<html lang="en">
<head><title>Guido's Personal Home Page</title></head>
<body>
  <h1>Guido van Rossum - Personal Home Page</h1>
  <p>I am the creator and <a href="http://www.python.org">Python</a> BDFL.</p>
  <p>Read my <a href="Resume.html">resume</a> and <a href="Publications.html">publications list</a>.</p>
</body>
</html>
"""

# Create a BeautifulSoup object from the HTML
soup = BeautifulSoup(sample_guido_html, 'html.parser')

# Prettify and extract title and text
print("--- Prettified HTML ---")
print(soup.prettify())

print("Title tag:", soup.title)
print("Extracted text:", soup.get_text().strip())

# Find all 'a' tags (hyperlinks) and print their href URLs
a_tags = soup.find_all('a')
print("\nExtracted Hyperlinks:")
for link in a_tags:
    print(link.get('href'))
```

### Chapter 1: Key Takeaways & Actionable Step-by-Step Process

#### Key Takeaways
* `pandas` (`read_csv`, `read_excel`) can stream tabular datasets directly from HTTP/HTTPS URLs into memory.
* `requests.get(url)` simplifies HTTP GET requests compared to `urllib` by automatically managing connections and decoding text via `r.text`.
* `BeautifulSoup` turns raw HTML strings into structured parse trees, making it straightforward to extract text (`.get_text()`) and hyperlinks (`.find_all('a')`).

#### Step-by-Step Process to Act On

3. **Parse with BeautifulSoup**: Initialize `soup = BeautifulSoup(html_doc, 'html.parser')`.
4. **Extract Target Elements**: Use `soup.find_all('a')` or tag selectors to loop through elements and extract `.get('href')` or `.get_text()`.

## Chapter 2: Interacting with APIs to Import Data from the Web

### 2.1 Introduction to APIs and JSONs
* **API (Application Programming Interface)**: A set of routines and protocols for building software and interacting with external web servers (e.g., OMDb Movie API, Wikipedia API).
* **JSON (JavaScript Object Notation)**: Standard human-readable format for transferring structured data across web APIs. Maps naturally to Python `dict` (key-value pairs) and `list` (arrays).
* **Loading Local JSON Files**: Use `import json` and `json.load(json_file)` inside a `with open(...)` context manager.

### 2.2 Fetching JSON from Web APIs with `requests`
* Send a GET request to the API URL (`r = requests.get(url, headers=headers)`).
* Decode the JSON response directly into a Python dictionary using `json_data = r.json()`.
* Navigate nested JSON dictionaries by chaining keys (e.g., `json_data['query']['pages']['24768']['extract']`).

```python
# Loading and exploring a local JSON file (a_movie.json)
import json

with open("datasets/a_movie.json", "r") as json_file:
    json_data = json.load(json_file)

print("Datatype of json_data:", type(json_data))
for k in json_data.keys():
    print(f"{k}: {json_data[k]}")
```

### Chapter 2: Key Takeaways & Actionable Step-by-Step Process

#### Key Takeaways
* JSON is the universal language of modern REST APIs and translates directly into Python dictionaries and lists.
* `json.load()` reads JSON from local files, while `response.json()` decodes JSON directly from an HTTP `requests` response.
* Deeply nested JSON payloads (like the Wikipedia API) are traversed by inspecting `.keys()` at each level.

#### Step-by-Step Process to Act On
1. **Build the API Endpoint URL**: Combine the base URL with required query parameters (e.g., `?apikey=...&t=social+network` or `params={...}`).
2. **Include Headers**: Pass a descriptive `User-Agent` header in `requests.get(url, headers=headers)` when required by providers like Wikipedia.
3. **Decode JSON**: Call `json_data = r.json()` to convert the payload into a Python dictionary.
4. **Traverse Keys**: Inspect `json_data.keys()` and index into nested dictionaries to extract the target fields.

## Chapter 3: Diving Deep into the Twitter API *(Added Context)*

> **Note**: Chapter 3 was listed in the course structure of your notebook (`tweets3.txt` dataset) and has been added below with full context and runnable code.

### 3.1 Twitter API Authentication & Streaming (`tweepy`)
* **Authentication**: Interacting with the Twitter/X API requires OAuth credentials (`consumer_key`, `consumer_secret`, `access_token`, `access_token_secret`) managed via the `tweepy` package.
* **Streaming vs. REST**: The Streaming API (`tweepy.StreamingClient` / `Stream`) captures real-time tweets matching filter keywords (`['clinton', 'trump', 'sanders', 'cruz']`), writing each tweet as a JSON line into a text file (`tweets3.txt`).

### 3.2 Parsing, Analyzing, and Plotting Tweet Data
* Read JSON-lines files line-by-line and parse each line with `json.loads(line)` into a list of dictionaries.
* Convert the list of tweet dictionaries into a `pandas` DataFrame (`df = pd.DataFrame(tweets_data, columns=['text', 'lang'])`).
* Use regular expressions (`re.search(word, text, re.IGNORECASE)`) to count keyword mentions across tweets and visualize frequencies with `seaborn` / `matplotlib`.

```python
# Loading, exploring, and plotting Twitter data from datasets/tweets3.txt
import json
import re
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Read tweets3.txt line by line and parse JSON
tweets_data = []
with open('datasets/tweets3.txt', 'r') as tweets_file:
    for line in tweets_file:
        tweet = json.loads(line)
        tweets_data.append(tweet)

print("First tweet keys:", tweets_data[0].keys())

# 2. Build DataFrame of tweet text and language
df_tweets = pd.DataFrame(tweets_data, columns=['text', 'lang'])
print("\nTweets DataFrame:")
print(df_tweets.head())

# 3. Define function to check if a word is in tweet text using regex
def word_in_text(word, text):
    word = word.lower()
    text = text.lower()
    match = re.search(word, text)
    if match:
        return True
    return False

# 4. Count mentions of each candidate
clinton, trump, sanders, cruz = 0, 0, 0, 0
for index, row in df_tweets.iterrows():
    clinton += word_in_text('clinton', row['text'])
    trump += word_in_text('trump', row['text'])
    sanders += word_in_text('sanders', row['text'])
    cruz += word_in_text('cruz', row['text'])

# 5. Plot keyword counts
cd = ['clinton', 'trump', 'sanders', 'cruz']
counts = [clinton, trump, sanders, cruz]
plt.figure(figsize=(6, 4))
sns.barplot(x=cd, y=counts)
plt.ylabel("Number of Tweets")
plt.title("Candidate Mentions in Tweet Stream")
plt.show()
```
